```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from typing import List
from pathlib import Path
from typing import Any
from datetime import datetime
from pandas import DataFrame
from pandas import Series

# Config
pd.options.display.float_format = '{:.3f}'.format

# Paths
home_path: str = str(Path.home())
mt5_relative_path: str = "\\AppData\\Roaming\\MetaQuotes\\Terminal\\Common\\Files\\csv_files"
csv_file_path: str = home_path + mt5_relative_path

# File Pattern
csv_file_pattern: str = "BLOCK_REVERSION_*.csv"

# Merge CSVs files
csv_df_list: List[DataFrame] = []
for csv_path in Path(csv_file_path).glob(csv_file_pattern):
    csv_df: DataFrame = pd.read_csv(csv_path, sep = ";")
    csv_df_list.append(csv_df)

# CSV to DataFrame
df_raw: DataFrame = pd.DataFrame()
if len(csv_df_list) > 0 :
    df_raw = pd.concat(csv_df_list, ignore_index=True)
    df_raw.drop_duplicates(inplace=True)

# Aditional Columns
df_raw["mfe_distance_blocks"] = np.where(
        (df_raw["block_size"] > 0), (df_raw["mfe_distance"] / df_raw["block_size"]), 0)

df_raw["mae_distance_blocks"] = np.where(
        (df_raw["block_size"] > 0), (df_raw["mae_distance"] / df_raw["block_size"]), 0)

# Dataset: 2026-01-05 - 2026-05-30
df_raw
```
